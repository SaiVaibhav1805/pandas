```python
df["month"] = df["t_date"].dt.month
def get_season(m):
    if m in [12,1,2]: return "Winter"
    elif m in [3,4,5]: return "Spring"
    elif m in [6,7,8]: return "Summer"
    else: return "Fall"

df["season"] = df["month"].apply(get_season)
pivot = pd.crosstab(df["services"], df["season"])
pivot.idxmax(axis=1)
```
